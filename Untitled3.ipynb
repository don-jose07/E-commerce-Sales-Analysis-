col names:Order_ID,Order_Date,	Customer,	City,Region,Category,	Product,	Quantity,	Unit_Price,	Discount_Percent,	Total_Sales,	Rating	,Payment_Method

*sales based on catagory


```python
category_sales = df.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)
display(category_sales)
```

### Observation: Sales by Category
From the above analysis, we can see the total sales for each product category. This helps in understanding which categories are performing well in terms of revenue.

```python
city_sales = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=False)
display(city_sales)
```

### Observation: Sales by City
This analysis reveals the total sales generated from each city. It can help identify the key geographical markets for the business.

```python
product_sales = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)
display(product_sales)
```

### Observation: Sales by Product
This analysis shows the total sales for each product. It helps in identifying which products are most popular and contribute significantly to overall revenue.

```python
payment_method_sales = df.groupby('Payment_Method')['Total_Sales'].sum().sort_values(ascending=False)
display(payment_method_sales)
```

### Observation: Sales by Payment Method
This analysis shows the total sales generated through each payment method. It helps in understanding which payment options are most frequently used by customers.

```python
total_sales = df['Total_Sales'].sum()
total_quantity_sold = df['Quantity'].sum()
number_of_orders = df['Order_ID'].nunique()
average_sales_per_order = df['Total_Sales'].mean()
highest_sale = df['Total_Sales'].max()
lowest_sale = df['Total_Sales'].min()

print(f"Total Sales: {total_sales:.2f}")
print(f"Total Quantity Sold: {total_quantity_sold}")
print(f"Number of Orders: {number_of_orders}")
print(f"Average Sales per Order: {average_sales_per_order:.2f}")
print(f"Highest Single Sale: {highest_sale:.2f}")
print(f"Lowest Single Sale: {lowest_sale:.2f}")
```

### Observation: Overall Sales Statistics
These statistics provide a general overview of the sales performance, including total revenue, volume of products sold, and the range of individual transaction values.

```python
high_value_orders = df[df['Total_Sales'] > 5000].sort_values(by='Total_Sales', ascending=False)
display(high_value_orders.head())
```

### Observation: High-Value Orders
This filtered view highlights orders with total sales exceeding a certain threshold, providing insight into significant transactions. Displaying the top few helps quickly identify major sales events.
